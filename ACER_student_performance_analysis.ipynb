```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

performance_data = pd.read_csv("student-mat.csv", sep=";")

performance_data.head()
```

```python
print("The Dataset shape:", performance_data.shape)

print("\nNames of Column:")
print(performance_data.columns.tolist())

print("\nInformation of Dataset:")
performance_data.info()
```

```python
print("The Missing values in each column:")
print(performance_data.isnull().sum())

print("\nThe Total missing values:")
print(performance_data.isnull().sum().sum())
```

```python
print("The Number of all duplicate rows:")
print(performance_data.duplicated().sum())
```

```python
performance_data.describe()
```

```python
print(performance_data["G3"].describe())

print("\nThe Final grade of frequencies:")
print(performance_data["G3"].value_counts().sort_index())
```

```python
performance_data["G3"].value_counts().sort_index().plot(kind="bar")

plt.xlabel("The Final Grade (G3)")
plt.ylabel("The Number of Students")
plt.title("The Distribution of Final Mathematics Grades")
plt.show()
```

```python
performance_data["performance_class"] = np.where(
    performance_data["G3"] >= 10,
    "Pass",
    "Fail"
)

print(performance_data["performance_class"].value_counts())
```

```python
performance_data["performance_class"].value_counts().plot(kind="bar")

plt.xlabel("The Performance Class")
plt.ylabel("The Number of Students")
plt.title("Pass and Fail Distribution")
plt.show()
```

```python
if "The performance_class" in performance_data.columns:
    performance_data = performance_data.drop(
        columns=["The performance_class"]
    )

X = performance_data.drop(
    columns=["G1", "G2", "G3", "performance_class"]
)

y = performance_data["performance_class"]

print("The Feature matrix shape:", X.shape)
print("The Target shape:", y.shape)

print("The Performance columns in X:",
      [col for col in X.columns if "performance" in col.lower()])
```

```python
X_encoded = pd.get_dummies(X, drop_first=True)

print("The Original number of the features:", X.shape[1])
print("The Features after encoding:", X_encoded.shape[1])

print("Target leakage check:")
print([
    col for col in X_encoded.columns
    if "performance" in col.lower()
])

X_encoded.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("The Training observations:", X_train.shape[0])
print("The Testing observations:", X_test.shape[0])
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

model_baseline = DummyClassifier(strategy="most_frequent")

model_baseline.fit(X_train, y_train)

predictions_baseline = model_baseline.predict(X_test)

accuracy_baseline = accuracy_score(y_test, predictions_baseline)

print("The Baseline accuracy:", accuracy_baseline)
```

```python
from sklearn.tree import DecisionTreeClassifier

tree_model = DecisionTreeClassifier(
    random_state=42,
    max_depth=5
)

tree_model.fit(X_train, y_train)

tree_predictions = tree_model.predict(X_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("The Decision Tree Accuracy:",
      accuracy_score(y_test, tree_predictions))

print("The Decision Tree Precision:",
      precision_score(y_test, tree_predictions, pos_label="Fail"))

print("The Decision Tree Recall:",
      recall_score(y_test, tree_predictions, pos_label="Fail"))

print("The Decision Tree F1:",
      f1_score(y_test, tree_predictions, pos_label="Fail"))

print("\nThe Confusion Matrix:")
print(confusion_matrix(y_test, tree_predictions))

print("\nThe Classification Report:")
print(classification_report(y_test, tree_predictions))
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.svm import SVC

svm_model = SVC(
    kernel="rbf",
    probability=True,
    random_state=42
)

svm_model.fit(X_train_scaled, y_train)

svm_predictions = svm_model.predict(X_test_scaled)
```

```python
print("SVM Accuracy:",
      accuracy_score(y_test, svm_predictions))

print("SVM Precision:",
      precision_score(y_test, svm_predictions, pos_label="Fail"))

print("SVM Recall:",
      recall_score(y_test, svm_predictions, pos_label="Fail"))

print("SVM F1:",
      f1_score(y_test, svm_predictions, pos_label="Fail"))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, svm_predictions))

print("\nClassification Report:")
print(classification_report(y_test, svm_predictions))
```

```python
comparison = pd.DataFrame({
    "Model": [
        "Decision Tree",
        "SVM"
    ],
    "Accuracy": [
        accuracy_score(y_test, tree_predictions),
        accuracy_score(y_test, svm_predictions)
    ],
    "Precision_Fail": [
        precision_score(y_test, tree_predictions, pos_label="Fail"),
        precision_score(y_test, svm_predictions, pos_label="Fail")
    ],
    "Recall_Fail": [
        recall_score(y_test, tree_predictions, pos_label="Fail"),
        recall_score(y_test, svm_predictions, pos_label="Fail")
    ],
    "F1_Fail": [
        f1_score(y_test, tree_predictions, pos_label="Fail"),
        f1_score(y_test, svm_predictions, pos_label="Fail")
    ]
})

comparison
```

```python
comparison.set_index("Model")[[
    "Accuracy",
    "Precision_Fail",
    "Recall_Fail",
    "F1_Fail"
]].plot(kind="bar")

plt.ylabel("Score")
plt.title("The Experimental Model Performance Comparison")
plt.ylim(0, 1)
plt.legend(loc="lower right")
plt.show()
```

```python
feature_importance = pd.DataFrame({
    "Feature": X_encoded.columns,
    "Importance": tree_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

feature_importance.head(10)
```

```python
top_features = feature_importance.head(10)

plt.barh(
    top_features["Feature"],
    top_features["Importance"]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("The Top 10 Decision Tree Feature Importances")
plt.gca().invert_yaxis()
plt.show()
```
