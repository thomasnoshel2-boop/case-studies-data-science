```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dropout_data = pd.read_csv("data.csv", sep=";")
```

```python
print(dropout_data.shape)
print(dropout_data.columns.tolist())

dropout_data.head()
```

```python
print("The Dataset shape:", dropout_data.shape)

print("\nThe Column names:")
print(dropout_data.columns.tolist())

print("\nThe Dataset information:")
dropout_data.info()
```

```python
print("The Missing values in each column:")
print(dropout_data.isnull().sum())

print("\nTotal missing values:")
print(dropout_data.isnull().sum().sum())
```

```python
print("The Number of duplicate rows:")
print(dropout_data.duplicated().sum())
```

```python
dropout_data.describe()
```

```python
print(dropout_data.columns.tolist())
```

```python
print("The Target classes:")
print(dropout_data["Target"].value_counts())

print("\nThe Target proportions:")
print(dropout_data["Target"].value_counts(normalize=True))
```

```python
dropout_data["Target"].value_counts().plot(kind="bar")

plt.xlabel("The Academic Outcome")
plt.ylabel("The Number of Students")
plt.title("The Distribution of Student Academic Outcomes")
plt.xticks(rotation=0)
plt.show()
```

```python
X_dropout = dropout_data.drop(columns=["Target"])
y_dropout = dropout_data["Target"]

print("The Feature matrix shape:", X_dropout.shape)
print("The Target shape:", y_dropout.shape)

print("\nThe Target classes:")
print(y_dropout.value_counts())
```

```python
X_dropout_encoded = pd.get_dummies(X_dropout, drop_first=True)

print("The Original number of features:", X_dropout.shape[1])
print("The Features after encoding:", X_dropout_encoded.shape[1])

X_dropout_encoded.head()
```

```python
from sklearn.model_selection import train_test_split

X_train_d, X_test_d, y_train_d, y_test_d = train_test_split(
    X_dropout_encoded,
    y_dropout,
    test_size=0.20,
    random_state=42,
    stratify=y_dropout
)

print("The Training observations:", X_train_d.shape[0])
print("The Testing observations:", X_test_d.shape[0])

print("\nThe Training target distribution:")
print(y_train_d.value_counts())

print("\nThe Testing target distribution:")
print(y_test_d.value_counts())
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score

baseline_dropout = DummyClassifier(
    strategy="most_frequent"
)

baseline_dropout.fit(X_train_d, y_train_d)

baseline_predictions_d = baseline_dropout.predict(X_test_d)

baseline_accuracy_d = accuracy_score(
    y_test_d,
    baseline_predictions_d
)

print("The Baseline Accuracy:", baseline_accuracy_d)
```

```python
from sklearn.tree import DecisionTreeClassifier

tree_dropout = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

tree_dropout.fit(X_train_d, y_train_d)

tree_predictions_d = tree_dropout.predict(X_test_d)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print("The Decision Tree Accuracy:",
      accuracy_score(y_test_d, tree_predictions_d))

print("The Decision Tree Macro Precision:",
      precision_score(
          y_test_d,
          tree_predictions_d,
          average="macro"
      ))

print("The Decision Tree Macro Recall:",
      recall_score(
          y_test_d,
          tree_predictions_d,
          average="macro"
      ))

print("The Decision Tree Macro F1:",
      f1_score(
          y_test_d,
          tree_predictions_d,
          average="macro"
      ))

print("\nThe Confusion Matrix:")
print(confusion_matrix(y_test_d, tree_predictions_d))

print("\nThe Classification Report:")
print(classification_report(y_test_d, tree_predictions_d))
```

```python
print(
    "The Dropout Precision:",
    precision_score(
        y_test_d,
        tree_predictions_d,
        labels=["Dropout"],
        average=None
    )[0]
)

print(
    "The Dropout Recall:",
    recall_score(
        y_test_d,
        tree_predictions_d,
        labels=["Dropout"],
        average=None
    )[0]
)

print(
    "The Dropout F1:",
    f1_score(
        y_test_d,
        tree_predictions_d,
        labels=["Dropout"],
        average=None
    )[0]
)
```

```python
feature_importance_d = pd.DataFrame({
    "Feature": X_dropout_encoded.columns,
    "Importance": tree_dropout.feature_importances_
})

feature_importance_d = feature_importance_d.sort_values(
    by="Importance",
    ascending=False
)

feature_importance_d.head(10)
```

```python
top_features_d = feature_importance_d.head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_features_d["Feature"],
    top_features_d["Importance"]
)

plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("The Top 10 Decision Tree Feature Importances - Student Dropout Dataset")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

scaler_d = StandardScaler()

X_train_d_scaled = scaler_d.fit_transform(X_train_d)
X_test_d_scaled = scaler_d.transform(X_test_d)

print("The Training data shape:", X_train_d_scaled.shape)
print("The Testing data shape:", X_test_d_scaled.shape)
```

```python
from sklearn.svm import SVC

svm_dropout = SVC(
    kernel="rbf",
    probability=True,
    random_state=42
)

svm_dropout.fit(X_train_d_scaled, y_train_d)

svm_predictions_d = svm_dropout.predict(X_test_d_scaled)

print("The SVM model training completed.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

print(
    "The SVM Accuracy:",
    accuracy_score(y_test_d, svm_predictions_d)
)

print(
    "The SVM Dropout Precision:",
    precision_score(
        y_test_d,
        svm_predictions_d,
        labels=["Dropout"],
        average=None
    )[0]
)

print(
    "The SVM Dropout Recall:",
    recall_score(
        y_test_d,
        svm_predictions_d,
        labels=["Dropout"],
        average=None
    )[0]
)

print(
    "The SVM Dropout F1:",
    f1_score(
        y_test_d,
        svm_predictions_d,
        labels=["Dropout"],
        average=None
    )[0]
)

print("\nThe Confusion Matrix:")
print(confusion_matrix(y_test_d, svm_predictions_d))

print("\nThe Classification Report:")
print(classification_report(y_test_d, svm_predictions_d))
```

```python
comparison_d = pd.DataFrame({
    "Model": [
        "Decision Tree",
        "SVM"
    ],

    "Accuracy": [
        accuracy_score(y_test_d, tree_predictions_d),
        accuracy_score(y_test_d, svm_predictions_d)
    ],

    "Precision_Dropout": [
        precision_score(
            y_test_d,
            tree_predictions_d,
            labels=["Dropout"],
            average=None
        )[0],

        precision_score(
            y_test_d,
            svm_predictions_d,
            labels=["Dropout"],
            average=None
        )[0]
    ],

    "Recall_Dropout": [
        recall_score(
            y_test_d,
            tree_predictions_d,
            labels=["Dropout"],
            average=None
        )[0],

        recall_score(
            y_test_d,
            svm_predictions_d,
            labels=["Dropout"],
            average=None
        )[0]
    ],

    "F1_Dropout": [
        f1_score(
            y_test_d,
            tree_predictions_d,
            labels=["Dropout"],
            average=None
        )[0],

        f1_score(
            y_test_d,
            svm_predictions_d,
            labels=["Dropout"],
            average=None
        )[0]
    ]
})

comparison_d
```

```python
comparison_d.set_index("Model")[[
    "Accuracy",
    "Precision_Dropout",
    "Recall_Dropout",
    "F1_Dropout"
]].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("The Model Performance Comparison - Student Dropout Dataset")
plt.ylabel("Score")
plt.xlabel("Model")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm_tree_d = confusion_matrix(
    y_test_d,
    tree_predictions_d
)

disp_tree_d = ConfusionMatrixDisplay(
    confusion_matrix=cm_tree_d,
    display_labels=tree_dropout.classes_
)

disp_tree_d.plot()

plt.title("Decision Tree Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
cm_svm_d = confusion_matrix(
    y_test_d,
    svm_predictions_d
)

disp_svm_d = ConfusionMatrixDisplay(
    confusion_matrix=cm_svm_d,
    display_labels=svm_dropout.classes_
)

disp_svm_d.plot()

plt.title("The SVM Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
final_comparison_d = pd.DataFrame({
    "Model": [
        "Baseline",
        "Decision Tree",
        "SVM"
    ],

    "Accuracy": [
        baseline_accuracy_d,
        accuracy_score(y_test_d, tree_predictions_d),
        accuracy_score(y_test_d, svm_predictions_d)
    ]
})

final_comparison_d
```

```python
final_comparison_d.plot(
    x="Model",
    y="Accuracy",
    kind="bar",
    legend=False,
    figsize=(8, 5)
)

plt.title("Baseline and Model Accuracy Comparison")
plt.xlabel("Model")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
